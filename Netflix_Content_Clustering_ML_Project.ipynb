## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# 1. Handle Missing Values
# ------------------------------

df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Not Available")
df["country"] = df["country"].fillna("Unknown")

# Fill rating with mode
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])

# ------------------------------
# 2. Clean and Convert date_added
# ------------------------------

# Remove extra spaces
df["date_added"] = df["date_added"].astype(str).str.strip()

# Convert to datetime safely
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")

# Drop rows where date conversion failed
df = df.dropna(subset=["date_added"])

# Extract year_added
df["year_added"] = df["date_added"].dt.year

# ------------------------------
# 3. Clean Duration Column
# ------------------------------

# Extract numeric part from duration
df["duration_int"] = df["duration"].str.extract("(\d+)").astype(int)

# Separate movie duration and TV seasons
df["movie_duration"] = np.where(df["type"] == "Movie", df["duration_int"], np.nan)
df["tv_seasons"] = np.where(df["type"] == "TV Show", df["duration_int"], np.nan)

# ------------------------------
# 4. Final Check
# ------------------------------

print("Remaining Missing Values:\n")
print(df.isnull().sum())

print("\nDataset Shape After Cleaning:", df.shape)

print("\nSample Data:")
df.head()
````

### What all manipulations have you done and insights you found?

Several preprocessing and data transformation steps were performed to prepare the dataset for analysis and clustering:

Missing values in director, cast, and country were filled with placeholders such as "Unknown" and "Not Available" to avoid unnecessary data loss.

Rows with invalid or missing date_added values were removed since they were minimal.

 This ensured the dataset remained complete and consistent without dropping significant records.

 This allowed separate analysis of movie length and TV show season count.

#### 1. State Your research hypothesis as a null hypothesis and alternate hypothesis.

H0 (Null Hypothesis): There is no significant difference in average duration between Movies and TV Shows.

H1 (Alternate Hypothesis): There is a significant difference in average duration between Movies and TV Shows.


#### 2. Perform an appropriate statistical test.

````python
# Perform Statistical Test to obtain P-Value
from scipy.stats import ttest_ind

movie_duration = df[df["type"] == "Movie"]["movie_duration"].dropna()
tv_duration = df[df["type"] == "TV Show"]["tv_seasons"].dropna()

t_stat, p_value = ttest_ind(movie_duration, tv_duration, equal_var=False)

print("T-Statistic:", t_stat)
print("P-Value:", p_value)
````

##### Which statistical test have you done to obtain P-Value?

An Independent Sample T-Test was performed using scipy.stats.ttest_ind() to obtain the p-value.


##### Why did you choose the specific statistical test?

The Independent Sample T-Test was chosen because we are comparing the mean duration of two independent groups — Movies and TV Shows.

The dependent variable (duration) is numerical, and the independent variable (type) is categorical with two groups. Therefore, a T-Test is the most appropriate statistical method to determine whether the difference in means is statistically significant.

### 4. Textual Data Preprocessing
(It's mandatory for textual dataset i.e., NLP, Sentiment Analysis, Text Clustering etc.)

#### 1. Expand Contraction

````python
# Expand Contraction
import re

contractions = {
    "can't": "cannot",
    "won't": "will not",
    "n't": " not",
    "'re": " are",
    "'s": " is",
    "'d": " would",
    "'ll": " will",
    "'t": " not",
    "'ve": " have",
    "'m": " am"
}

def expand_contractions(text):
    for key, value in contractions.items():
        text = re.sub(key, value, text)
    return text

df["description"] = df["description"].apply(expand_contractions)
````

#### 2. Lower Casing

````python
# Lower Casing
# Convert text to lowercase

df["description"] = df["description"].str.lower()
````

#### 3. Removing Punctuations

````python
# Remove Punctuations
import string

def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

df["description"] = df["description"].apply(remove_punctuation)
````

#### 4. Removing URLs & Removing words and digits contain digits.

````python
# Remove URLs & Remove words and digits contain digits
import re

def clean_text(text):
    # 1️. Remove URLs
    text = re.sub(r'http\S+|www\S+|https\S+', '', text)

    # 2️. Remove words containing digits
    text = re.sub(r'\w*\d\w*', '', text)

    return text

df["description"] = df["description"].apply(clean_text)

df["description"].head()
````

#### 5. Removing Stopwords & Removing White spaces

````python
# Remove Stopwords
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

stop_words = set(stopwords.words('english'))

df["description"] = df["description"].apply(
    lambda x: " ".join([word for word in x.split() if word not in stop_words])
)
````

````python
# Remove White spaces
import re

df["description"] = df["description"].apply(
    lambda x: re.sub(r"\s+", " ", x).strip()
)
````

#### 6. Rephrase Text

````python
# Rephrase Text
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')

lemmatizer = WordNetLemmatizer()

df["description"] = df["description"].apply(
    lambda x: " ".join([lemmatizer.lemmatize(word) for word in x.split()])
)
````

#### 7. Tokenization

````python
# Tokenization

import nltk

nltk.download('punkt')
nltk.download('punkt_tab')

from nltk.tokenize import word_tokenize
nltk.download('punkt')

df["tokens"] = df["description"].apply(word_tokenize)

df["tokens"].head()
````

#### 8. Text Normalization

````python
# Normalizing Text (i.e., Stemming, Lemmatization etc.)
def normalize_text(text):
    text = text.lower()
    text = re.sub(r"\s+", " ", text)
    return text.strip()

df["description"] = df["description"].apply(normalize_text)
````

##### Which text normalization technique have you used and why?

In this project, I applied several text normalization techniques to clean and standardize the textual data before using it for clustering.

First, I converted all text to lowercase. This ensures that words like “Movie” and “movie” are treated as the same word, which helps reduce duplication in the vocabulary.

Next, I removed punctuation marks such as commas, periods, and special characters. These do not add meaningful information for clustering and only increase noise in the dataset.

I also removed URLs from the text. Links do not contribute to understanding the content of a movie or TV show, so keeping them would not help the model.

#### 2. Cross- Validation & Hyperparameter Tuning

````python
# ML Model - 1 Implementation with hyperparameter optimization techniques (i.e., GridSearch CV, RandomSearch CV, Bayesian Optimization etc.)

# Fit the Algorithm

# Predict on the model

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np
import pandas as pd

# Define parameter grid
param_grid = {
    "n_clusters": [2, 3, 4, 5, 6, 7, 8],
    "init": ["k-means++", "random"],
    "max_iter": [200, 300, 400]
}

best_score = -1
best_params = {}
best_model = None

# Manual Grid Search using Silhouette Score
for n in param_grid["n_clusters"]:
    for init in param_grid["init"]:
        for max_iter in param_grid["max_iter"]:

            model = KMeans(
                n_clusters=n,
                init=init,
                max_iter=max_iter,
                random_state=42
            )

            labels = model.fit_predict(X_scaled)
            score = silhouette_score(X_scaled, labels)

            if score > best_score:
                best_score = score
                best_params = {
                    "n_clusters": n,
                    "init": init,
                    "max_iter": max_iter
                }
                best_model = model

# Fit best model
best_model.fit(X_scaled)
final_labels = best_model.predict(X_scaled)

print("Best Parameters:", best_params)
print("Best Silhouette Score:", best_score)

````
